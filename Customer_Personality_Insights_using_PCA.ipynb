```python
df["Income"] = df["Income"].fillna(df["Income"].median())
```

```python
df["Income"].isnull().sum()
```

```python
df["Dt_Customer"] = pd.to_datetime(

    df["Dt_Customer"],

    dayfirst=True

)
```

```python
df["Dt_Customer"].head()
```

```python
current_year = 2026
df["Age"] = current_year - df["Year_Birth"]
```

```python
df[["Year_Birth","Age"]].head()
```

```python
df["Total_Children"] = df["Kidhome"] + df["Teenhome"]
```

```python
df["Total_Spending"] = (

      df["MntWines"]

    + df["MntFruits"]

    + df["MntMeatProducts"]

    + df["MntFishProducts"]

    + df["MntSweetProducts"]

    + df["MntGoldProds"]

)
```

```python
df["Total_Spending"].head()
```

```python
df["Total_Purchases"] = (

      df["NumWebPurchases"]

    + df["NumCatalogPurchases"]

    + df["NumStorePurchases"]

)
```

```python
df["Customer_Since"] = (

    current_year -

    df["Dt_Customer"].dt.year

)
```

```python
df[

    [

        "Age",

        "Total_Children",

        "Total_Spending",

        "Total_Purchases",

        "Customer_Since"

    ]

].head()
```

```python
df.drop(

    columns=[

        "ID",

        "Year_Birth",

        "Dt_Customer"

    ],

    inplace=True

)
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
le = LabelEncoder()
```

```python
df["Education"] = le.fit_transform(df["Education"])
```

```python
df["Marital_Status"] = le.fit_transform(df["Marital_Status"])
```

```python
df.head()
```

```python
# Target Variable

y = df["Response"]

# Features

X = df.drop("Response", axis=1)
```

```python
print(X.shape)

print(y.shape)
```

```python
X.columns
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
```

```python
X_scaled = scaler.fit_transform(X)
```

```python
print(X_scaled[:5])
```

```python
X_scaled = pd.DataFrame(

    X_scaled,

    columns=X.columns

)
```

```python
X_scaled.head()
```

```python
from sklearn.decomposition import PCA
```

```python
pca = PCA(n_components=2)
```

```python
X_pca = pca.fit_transform(X_scaled)
```

```python
print(X_pca.shape)
```

```python
pca_df = pd.DataFrame(

    X_pca,

    columns=["PC1","PC2"]

)
```

```python
pca_df.head()
```

```python
print(pca.explained_variance_ratio_)
```

```python
print(sum(pca.explained_variance_ratio_))
```

```python
for i, variance in enumerate(pca.explained_variance_ratio_):

    print(f"PC{i+1} : {variance:.3f}")
```

```python
print(pca.components_)
```

```python
print(pca.explained_variance_)
```

```python
print(pca.singular_values_)
```

```python
print("Original Shape :", X_scaled.shape)

print("Reduced Shape :", X_pca.shape)
```

```python
print("="*50)

print("Original Features :", X_scaled.shape[1])

print("Reduced Features :", X_pca.shape[1])

print("Variance Preserved :", round(sum(pca.explained_variance_ratio_)*100,2),"%")

print("="*50)
```

```python
# PC1 me har feature ka contribution

feature_importance = pd.DataFrame(

    pca.components_.T,

    columns=["PC1","PC2"],

    index=X.columns

)

feature_importance
```

```python
# PC1 ke top contributing features

feature_importance["PC1"].abs().sort_values(

    ascending=False

).head(10)
```

```python
# ==========================================================
# Step 71 : PC1 vs PC2 Scatter Plot
# ==========================================================

# Figure size set kar rahe hain
plt.figure(figsize=(10,7))

# Scatter plot bana rahe hain
plt.scatter(

    pca_df["PC1"],          # X-axis -> First Principal Component

    pca_df["PC2"],          # Y-axis -> Second Principal Component

    alpha=0.7,              # Points ko thoda transparent banaya hai

    edgecolor="black"       # Har point ka border black hoga

)

# Graph ka title
plt.title("PCA Projection (PC1 vs PC2)", fontsize=16)

# X-axis label
plt.xlabel("Principal Component 1 (PC1)", fontsize=12)

# Y-axis label
plt.ylabel("Principal Component 2 (PC2)", fontsize=12)

# Grid add kar rahe hain
plt.grid(True)

# Graph show karo
plt.show()
```
